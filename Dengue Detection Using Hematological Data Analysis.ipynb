OUTLIER HANDALING

```python
# Function to detect and cap outliers using IQR
def handle_outliers(df):
    df_clean = df.copy()
    summary = []

    for col in df.select_dtypes(include=['int64', 'float64']).columns:

        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        # detect outliers
        outliers = df[(df[col] < lower) | (df[col] > upper)][col]

        # cap (fix) outliers
        df_clean[col] = df[col].clip(lower, upper)

        # store info
        summary.append({
            'Column': col,
            'Outliers': len(outliers),
            'Max': outliers.max() if len(outliers) > 0 else None,
            'Min': outliers.min() if len(outliers) > 0 else None
        })

    return df_clean, pd.DataFrame(summary)
```

```python
df_clean, report = handle_outliers(df)

print(report)
```

```python
from scipy.stats import ttest_ind
# Split into two groups
positive = df[df['Result'] == 'positive']
negative = df[df['Result'] == 'negative']

# Select numeric columns only
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

# Run t-test for all variables
results = []

for col in numeric_cols:
    t_stat, p_value = ttest_ind(
        positive[col],
        negative[col],
        equal_var=False  # Welch’s t-test
    )

    results.append({
        'Variable': col,
        'Mean_Positive': positive[col].mean(),
        'Mean_Negative': negative[col].mean(),
        'T_stat': t_stat,
        'P_value': p_value,
        'Significant (p<0.05)': p_value < 0.05
    })

# Convert to DataFrame and sort
results_df = pd.DataFrame(results).sort_values(by='P_value')

# Show results
results_df
```

Independent sample t-tests were conducted to compare hematological parameters between dengue positive and negative patients. Significant differences were observed in variables such as neutrophils, lymphocytes, and platelet count (p < 0.05), suggesting their strong association with dengue infection. In contrast, variables such as hemoglobin and PDW showed no significant differences (p > 0.05), indicating limited diagnostic relevance.
